## 🌐 Method 2: Load CSV from GitHub

✔ Best for assignments  
✔ Shareable and reproducible  

### Steps:
1. Open CSV in GitHub
2. Click **Raw**
3. Copy URL

````python
import pandas as pd

url = "https://raw.githubusercontent.com/Dong2Yo/Dataset/refs/heads/main/RFM_Asia.csv"

data = pd.read_csv(url)
data.head()
````

⚠️ Common Issues:
- "Link too long" → NOT a raw link
- Must start with:
  https://raw.githubusercontent.com/
