## Business case setup
Northstar Coffee tracks net sales for three stores across seven days. Management wants a compact workflow to:
1. inspect the structure of the data;
2. calculate tax-adjusted and percentage-change values;
3. identify strong or weak store-days;
4. summarize performance by store and by day;
5. avoid accidental changes to source data.

Rows represent stores and columns represent Monday through Sunday.

```python
import numpy as np
import matplotlib.pyplot as plt

stores = np.array(["Downtown", "Campus", "Airport"])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

sales = np.array([
    [820, 910, 875, 940, 1100, 1350, 1200],
    [640, 700, 720, 760, 880,  930,  810],
    [990, 960, 1020, 1080, 1250, 1480, 1410]
], dtype=float)

sales
```

## 1. Arrays, dimensions, shape, and type
A NumPy `ndarray` stores a dense collection of values with a common data type. The shape describes the length of each axis.

### Skills
- interpret rows and columns in context;
- inspect `ndim`, `shape`, `size`, and `dtype`;
- connect dimensions to business meaning.

### Common issues
- confusing `size` with `shape`;
- assuming rows and columns have meaning without documenting them;
- mixing text and numbers, which can promote all values to a text-compatible type.

```python
print("Dimensions:", sales.ndim)
print("Shape:", sales.shape)
print("Element count:", sales.size)
print("Data type:", sales.dtype)
print("Stores:", len(stores), "Days:", len(days))
```

## 2. Creating and combining arrays
Useful constructors include `np.zeros`, `np.ones`, `np.arange`, and random-number generators. `astype()` converts data to another type. `np.concatenate()` joins compatible arrays along an existing axis.

### Skills
- choose a constructor based on intent;
- create matching arrays with `zeros_like` or `ones_like`;
- convert types deliberately;
- concatenate arrays along the correct axis.

### Common issues
- expecting `arange` to include its stop value;
- combining arrays with incompatible non-concatenation dimensions;
- converting a float array to integer and losing decimal portions.

```python
targets = np.full_like(sales, 1000.0)
variance = sales - targets
week_numbers = np.arange(1, 8)
closed_flags = np.zeros_like(sales, dtype=bool)

print("Week numbers:", week_numbers)
print("Variance shape:", variance.shape)
print("Any closures recorded?", closed_flags.any())
```

```python
next_monday = np.array([[860.0], [670.0], [1010.0]])
extended_sales = np.concatenate([sales, next_monday], axis=1)
print("Original shape:", sales.shape)
print("Extended shape:", extended_sales.shape)
```

## 3. Reshaping and element-count rules
`reshape()` changes the arrangement, not the number of elements. One dimension may be `-1`, allowing NumPy to infer its length.

### Skills
- calculate whether a requested shape is possible;
- flatten or restructure data;
- use `-1` appropriately.

### Common issues
- requesting a shape with a different element count;
- losing the documented business meaning of axes;
- assuming a reshaped result is always an independent copy.

```python
flat_sales = sales.reshape(-1)
print("Flat shape:", flat_sales.shape)

three_by_seven = flat_sales.reshape(3, -1)
print("Reconstructed shape:", three_by_seven.shape)
print("Same values:", np.array_equal(sales, three_by_seven))
```

## 4. Views and copies
Many slices are views, meaning they share data with the original array. Use `.copy()` when independent data is required.

### Issue demonstration
The first example changes the source array through a slice. The second protects the source by making a copy.

```python
demo = sales.copy()
weekend_view = demo[:, 5:]
weekend_view[0, 0] = -999
print("Source changed through view:", demo[0, 5])

safe_demo = sales.copy()
weekend_copy = safe_demo[:, 5:].copy()
weekend_copy[0, 0] = -999
print("Independent copy changed:", weekend_copy[0, 0])
print("Protected source value:", safe_demo[0, 5])
```

## 5. Vectorized operations and universal functions
Vectorization applies operations to arrays without writing an explicit Python loop for every element. Arithmetic is generally element-wise. Comparisons produce Boolean arrays. Universal functions, or ufuncs, include `np.sqrt`, `np.exp`, and `np.isnan`.

### Skills
- apply one formula across a full array;
- aggregate along an axis;
- build Boolean masks;
- use ufuncs for element-wise calculations.

### Common issues
- confusing element-wise multiplication `*` with matrix multiplication `@`;
- using the wrong aggregation axis;
- performing an in-place operation that unexpectedly changes the original array.

```python
tax_rate = 0.05
sales_with_tax = sales * (1 + tax_rate)
store_totals = sales.sum(axis=1)
day_totals = sales.sum(axis=0)
store_averages = sales.mean(axis=1)

print("Store totals:", store_totals)
print("Day totals:", day_totals)
print("Store averages:", np.round(store_averages, 2))
print("Square roots of first store sales:", np.round(np.sqrt(sales[0]), 2))
```

## 6. Indexing, slicing, and Boolean masks
NumPy uses zero-based indexing. In two dimensions, the first index selects a row and the second selects a column. Slice syntax is `start:stop:step`, with the stop excluded.

### Skills
- retrieve a single value, row, column, or rectangular region;
- select non-adjacent rows with an index list;
- filter values using a Boolean condition.

### Common issues
- off-by-one errors;
- reversing row and column positions;
- expecting a Boolean mask to preserve the original two-dimensional shape when extracting values.

```python
print("Top-left value:", sales[0, 0])
print("First store:", sales[0, :])
print("Friday across stores:", sales[:, 4])
print("Weekdays only:")
print(sales[:, :5])
print("Downtown and Airport:")
print(sales[[0, 2], :])

meets_target = sales >= 1000
print("Meets-target mask:")
print(meets_target)
print("Values meeting target:", sales[meets_target])
```
