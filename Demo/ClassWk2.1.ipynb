## 8. Dictionaries

Lists store values by position.

Dictionaries store values using labels.

This looks much more like a business record.

```python
customer = {
    "customer_id": 1001,
    "name": "Sarah Johnson",
    "city": "Edmonton",
    "member": True
}
```

```python
print(customer["name"])

print(customer["city"])
```

```python
customer["city"] = "Calgary"

print(customer)
```

```python
customer["member"] = False

print(customer)
```

Why is this:
