## 7. Business Files

Businesses store data in files every day.

Examples include:

- CSV files
- Excel files
- Text files

Let's create a simple sales file.

```python
sales_text = """
Laptop,1200
Monitor,350
Keyboard,75
Mouse,45
"""
```

```python
with open("sales.txt", "w") as file:
    file.write(sales_text)

print("Sales file created.")
```

Our file now contains product sales information that can be read by Python.

### Opening a File

Before Python can read a file, it must open it.

```python
file = open("sales.txt", "r")

print(file)
```

### Reading File Contents

Let's read the entire file.

```python
file = open("sales.txt", "r")

contents = file.read()

print(contents)
```

Notice that the contents are returned as a single string.

### Reading One Record at a Time

Business reports often process one transaction at a time.

```python
file = open("sales.txt", "r")

for line in file:
    print(line)
```

Each line represents a business record.

### How Many Products Are Stored?

Managers often ask:

"How many records do we have?"

```python
file = open("sales.txt", "r")

count = 0

for line in file:
    count += 1

print("Number of products:", count)
```

### Searching for Specific Information

Suppose the manager wants to find information about laptops.

```python
file = open("sales.txt", "r")

for line in file:
    
    if line.startswith("Laptop"):
        print(line)
```

## 8. Mini Analyst Project

You are preparing a report for management.

Tasks:

1. Store monthly sales in an array
2. Calculate average sales
3. Use a function to calculate revenue
4. Read product information from a file

Can you combine everything learned today?

```python
import numpy as np

monthly_sales = np.array([
    120,
    150,
    135,
    170
])

def calculate_revenue(price, quantity):
    return price * quantity

average_sales = monthly_sales.mean()

revenue = calculate_revenue(1200, 3)

print("Average Sales:", average_sales)

print("Revenue:", revenue)
```
