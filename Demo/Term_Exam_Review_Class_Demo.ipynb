- `int` represents whole numbers.
- `float` represents numbers with decimals.

## Strings

# 25–40 Minutes: Loops

**Class focus:** Trace each loop by hand before executing it. Ask students to identify the changing variable and stopping condition.

## `for` loops
Use a `for` loop to repeat once for each item in a sequence.

## `while` loops
Use a `while` loop while a condition remains `True`. Make sure something eventually changes the condition to avoid an infinite loop.

## Important keywords
- `range(start, stop, step)` generates integer values; `stop` is excluded.
- `break` exits the loop.
- `continue` skips to the next iteration.
- An **accumulator** stores a running result, such as a total.


````python
for number in range(1, 5):
    print(number)
````

````python
total = 0
for number in [2, 4, 6]:
    total += number
    #total = number + total
print(total)
````

````python
count = 3
while count > 0:
    print(count)
    count -= 1
````

````python
## This is an infinite loop. PLEASE DO NOT EXECUTE.
count = 3
while count > 0:
    print(count)
    count += 1
````

### Trace the loop

# 40–55 Minutes: Introduction to Pandas

**Class focus:** Create and inspect a DataFrame, then practise selecting columns and filtering rows.

## Core objects
- A **Series** is one-dimensional.
- A **DataFrame** is a two-dimensional table with rows and columns.
- Import pandas with `import pandas as pd`.
- Read a CSV file with `pd.read_csv()`.

## Essential inspection tools
- `.head()` and `.tail()` preview rows.
- `.shape` gives `(rows, columns)`.
- `.columns` lists column names.
- `.dtypes` shows each column's data type.
- `.info()` summarizes structure, non-null counts, and data types.

````python
import pandas as pd

students = pd.DataFrame({
    "name": ["Ava", "Ben", "Chen", "Dina"],
    "score": [82, 74, 91, 88],
    "program": ["DMIT", "DMIT", "BAIST", "BAIST"]
})

students
````

## Selecting data

- Multiple columns as a DataFrame: `df[["col1", "col2"]]`
- Label-based selection: `.loc[]`
- Position-based selection: `.iloc[]`
- Filter rows with a Boolean condition.


````python
students["score"]
````

````python
students[["name", "score"]]
````

````python
students.loc[students["score"] >= 85, ["name", "score"]]
````

# 55–75 Minutes: Data Cleaning and Preparation with Pandas

**Class focus:** Spend extra time distinguishing true missing values from invalid placeholders such as `?`.

## Missing and invalid values
- `NaN` represents a missing value in many pandas operations.
- `.isna()` or `.isnull()` detects missing values.
- `.notna()` detects non-missing values.
- `.dropna()` removes rows or columns containing missing values.
- `.fillna()` replaces missing values.
- A placeholder such as `?`, `N/A`, or `unknown` may be stored as ordinary text unless it is converted or identified when reading the data.

> **Key distinction:** A missing value and an invalid placeholder are not automatically the same thing.


````python
dirty = pd.DataFrame({
    "name": [" Ava ", "Ben", "Chen", "Ben"],
    "age": [20, None, "?", None],
    "city": ["Edmonton", "edmonton", "Calgary", "edmonton"]
})

dirty
````

````python
# Replace the invalid placeholder with a true missing value
clean = dirty.replace("?", pd.NA)

# Standardize text
clean["name"] = clean["name"].str.strip()
clean["city"] = clean["city"].str.title()

clean
````

````python
# Count missing values by column
clean.isna().sum()
````

````python
# Convert age to a numeric data type; invalid values become NaN
clean["age"] = pd.to_numeric(clean["age"], errors="coerce")
clean.dtypes
````

````python
# Find and remove duplicate rows
print(clean.duplicated().sum())
clean_no_duplicates = clean.drop_duplicates()
clean_no_duplicates
````

## Common cleaning operations
- Rename columns: `df.rename(columns={"old": "new"})`
- Change data type: `.astype()` or `pd.to_numeric()`
- Replace values: `.replace()`
- Standardize strings: `.str.strip()`, `.str.lower()`, `.str.upper()`, `.str.title()`
- Detect duplicates: `.duplicated()`
- Remove duplicates: `.drop_duplicates()`

# 75–90 Minutes: Data Wrangling with Pandas

**Class focus:** Connect each operation to a business question: filter, sort, summarize, combine, or reshape.

Data wrangling means reshaping, combining, filtering, sorting, summarizing, and transforming data for analysis.

## Common operations
- Sort: `.sort_values()`
- Filter: Boolean conditions
- Create a column: `df["new"] = ...`
- Group and summarize: `.groupby().agg()`
- Combine tables by a key: `pd.merge()`
- Stack tables: `pd.concat()`
- Reshape summaries: `.pivot_table()`


````python
sales = pd.DataFrame({
    "region": ["North", "North", "South", "South"],
    "product": ["Coffee", "Tea", "Coffee", "Tea"],
    "units": [10, 8, 12, 7],
    "price": [4.0, 3.0, 4.0, 3.0]
})

sales["revenue"] = sales["units"] * sales["price"]
sales
````

````python
# Sort from highest to lowest revenue
sales.sort_values("revenue", ascending=False)
````

````python
# Group and summarize
sales.groupby("region", as_index=False).agg(
    total_units=("units", "sum"),
    total_revenue=("revenue", "sum")
)
````

````python
# Filter with two conditions; use parentheses around each condition
sales[(sales["region"] == "South") & (sales["units"] >= 10)]
````

````python
products = pd.DataFrame({
    "product": ["Coffee", "Tea"],
    "category": ["Hot Drink", "Hot Drink"]
})

pd.merge(sales, products, on="product", how="left")
````

````python
sales.pivot_table(
    index="region",
    columns="product",
    values="revenue",
    aggfunc="sum"
)
````

### `merge` versus `concat`
- Use `merge` to combine tables using matching key columns.
- Use `concat` to stack or align objects along rows or columns.
