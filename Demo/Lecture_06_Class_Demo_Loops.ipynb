## 3. Counters and accumulators

- A **counter** tracks how many times something happens.
- An **accumulator** builds a running total.
- Both must be initialized before the loop.

```python
order_count = 0
subtotal_total = 0

for order in orders:
    order_count += 1
    #order_count = order_count + 1
    subtotal_total += order["subtotal"]

print("Order records:", order_count)
print("Raw subtotal total:", round(subtotal_total, 2))
```

## 4. Conditions inside a loop

A loop repeats the work. An `if` statement chooses what happens during each repetition.

```python
for order in orders:
    if order["subtotal"] >= 100:
        status = "high-value"
    elif order["subtotal"] >= 0:
        status = "standard"
    else:
        status = "invalid"
    print(order["order_id"], status)
```

## 5. Use `continue` to skip invalid data

`continue` immediately starts the next iteration. Invalid records are reported but excluded from valid totals.

```python
valid_count = 0
valid_subtotal_total = 0

for order in orders:
    if order["subtotal"] < 0 or order["items"] < 1:
        print(order["order_id"], "skipped: invalid data")
        continue
    valid_count += 1
    valid_subtotal_total += order["subtotal"]

print("Valid orders:", valid_count)
print("Valid subtotal total:", round(valid_subtotal_total, 2))
```

## 6. Use `break` to stop a search

`break` exits the nearest loop. This is useful when only the first match is needed.

```python
for order in orders:
    if order["subtotal"] >= 120:
        print("First order requiring review:", order["order_id"])
        break
else:
    print("No order requires review")
```

## 7. Nested loops

A nested loop repeats an inner loop for every iteration of an outer loop. Use this pattern only when the problem has two genuine levels of repetition.

```python
weekly_orders = {
    "Monday": ["NC-101", "NC-102"],
    "Tuesday": ["NC-103", "NC-104", "NC-105"],
}

for day, order_ids in weekly_orders.items():
    print(day)
    for order_id in order_ids:
        print("  ", order_id)
```

## 8. A safe `while` loop

A `while` loop repeats while its condition is true. The control variable must change so the loop can end.

```python
index = 0
while index < len(orders):
    order = orders[index]
    print(index, order["order_id"])
    index += 1
```

## 9. Integrated business-case application

### Rules
- An order is invalid when the subtotal is negative or the item count is less than 1.
- A loyalty member receives a 10% discount when the subtotal is at least `$50`.
- Another valid order receives a 5% discount when the subtotal is at least `$80`.
- A `$5` delivery fee applies when the discounted subtotal is below `$40`.
- A valid final total of at least `$100` requires manager review.

```python
def calculate_order(order):
    subtotal = order["subtotal"]
    items = order["items"]
    loyalty = order["loyalty"]

    if subtotal < 0 or items < 1:
        return None

    if loyalty and subtotal >= 50:
        discount_rate = 0.10
    elif subtotal >= 80:
        discount_rate = 0.05
    else:
        discount_rate = 0.0

    discount = subtotal * discount_rate
    discounted_subtotal = subtotal - discount
    delivery_fee = 5.0 if discounted_subtotal < 40 else 0.0
    final_total = discounted_subtotal + delivery_fee
    review_status = "manager review" if final_total >= 100 else "standard"

    return {
        "discount_rate": discount_rate,
        "discount": round(discount, 2),
        "delivery_fee": delivery_fee,
        "final_total": round(final_total, 2),
        "review_status": review_status,
    }
```

```python
results = []
invalid_count = 0

for order in orders:
    calculation = calculate_order(order)
    if calculation is None:
        invalid_count += 1
        #invalid_count = invalid_count + 1
        print(order["order_id"], "invalid")
        continue
    result = {"order_id": order["order_id"], **calculation}
    results.append(result)
    print(result)

revenue = sum(result["final_total"] for result in results)
average_order = revenue / len(results) if results else 0

print("\nValid orders:", len(results))
print("Invalid orders:", invalid_count)
print("Daily revenue:", round(revenue, 2))
print("Average valid order:", round(average_order, 2))
```

## 10. Visualize the loop results

The loop prepared `results`. The chart communicates the final totals by order.

```python
order_ids = [result["order_id"] for result in results]
final_totals = [result["final_total"] for result in results]

plt.figure(figsize=(8, 4))
plt.bar(order_ids, final_totals)
plt.axhline(100, color="red", linestyle="--", label="Manager-review threshold")
plt.title("Final Total by Valid Order")
plt.xlabel("Order ID")
plt.ylabel("Final total ($)")
plt.legend()
plt.tight_layout()
plt.show()
```
