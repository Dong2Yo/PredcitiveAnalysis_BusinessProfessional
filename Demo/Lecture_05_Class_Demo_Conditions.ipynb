## 2. Conditional branches
Python checks branches from top to bottom. In an `if`/`elif`/`else` chain, only the first true branch runs.

### Skills
- use indentation to define a block;
- order conditions from invalid or most specific to general;
- ensure every relevant situation has a clear outcome.

### Common issues
- inconsistent indentation;
- putting a broad condition before a more specific one;
- using separate `if` statements when the branches should be mutually exclusive.

```python
def classify_order(order):
    if order["subtotal"] < 0 or order["items"] < 1:
        return "invalid"
    elif order["subtotal"] >= 100:
        return "large"
    elif order["subtotal"] >= 50:
        return "medium"
    else:
        return "small"

for order in orders:
    print(order["order_id"], classify_order(order))
```

```python
categories = ["small", "medium", "large", "invalid"]
category_counts = []

for category in categories:
    count = 0
    for order in orders:
        if classify_order(order) == category:
            count += 1
    category_counts.append(count)

plt.figure(figsize=(7, 4))
plt.bar(categories, category_counts)
plt.title("Northstar Coffee Order Classifications")
plt.xlabel("Order category")
plt.ylabel("Number of orders")
plt.show()
```
