# 1. Understanding Google Colab

Google Colab notebooks typically contain two types of cells.

## Markdown Cells

Markdown cells contain:

- Titles
- Explanations
- Notes
- Documentation

Markdown cells do NOT execute Python code.

Think of Markdown cells as the "report" portion of your notebook.

---

## Code Cells

Code cells contain Python instructions.

Code cells can be executed by clicking the Run button.

Think of Code cells as the "work" portion of your notebook.

---

## Why Do We Need Both?

Professional analysts don't just write code.

They also document:

- What they are doing
- Why they are doing it
- What conclusions they found

A good notebook contains both Markdown and Code.

# 8. Building a Daily Sales Report

We will now combine:

✅ Variables

✅ Strings

✅ Numbers

✅ Expressions

✅ Statements

into a simple business application.

```python
store_name = "Downtown Coffee"

manager = "Sarah"

coffee_sold = 85
coffee_price = 5.75

muffin_sold = 40
muffin_price = 3.25

coffee_revenue = coffee_sold * coffee_price

muffin_revenue = muffin_sold * muffin_price

total_revenue = coffee_revenue + muffin_revenue

print("DAILY SALES REPORT")
print("----------------------")

print("Store:", store_name)

print("Manager:", manager)

print("Coffee Revenue:", coffee_revenue)

print("Muffin Revenue:", muffin_revenue)

print("Total Revenue:", total_revenue)
```
