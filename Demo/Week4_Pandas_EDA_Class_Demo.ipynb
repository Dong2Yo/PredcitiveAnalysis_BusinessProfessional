## 2. Introduction to Pandas

### 2.1 What is Pandas?

**Pandas** is one of the most widely used Python libraries for data analysis and data manipulation. It provides powerful tools for working with structured data such as spreadsheets, CSV files, and databases.

Pandas allows analysts to:

- Load data from files and databases
- Clean and prepare data
- Explore and summarize information
- Filter and transform records
- Perform calculations and analysis
- Export results for reporting

Pandas is built on top of **NumPy** and is designed to make working with data simple, efficient, and intuitive.

---

### 2.2 Why Do We Use Pandas?

Imagine a healthcare organization collects hundreds of patient encounters every week. An analyst may want to answer questions such as:

- Which clinic has the longest average wait time?
- Which service line has the highest average cost?
- How many encounters were completed this month?
- Are there missing values in the dataset?

Answering these questions manually can be time-consuming. Pandas provides tools that allow us to analyze large amounts of data quickly and accurately.

## 6. Selecting with `[]`, `.loc[]`, and `.iloc[]`
- `[]` supports direct column selection and simple Boolean filtering.
- `.loc[]` selects by labels or conditions.
- `.iloc[]` selects by zero-based positions.

````python
# []: select one Series and then a multi-column DataFrame.
display(df["Service_Line"].head())
display(df[["Encounter_ID", "Facility", "Service_Line", "Wait_Time_Min"]].head())
````

````python
# .loc[]: label-based selection; the ending label is included.
display(df_by_encounter.loc[
    "E0005":"E0010",
    ["Facility", "Service_Line", "Wait_Time_Min", "Cost_CAD"]
])
````

````python
# .iloc[]: position-based selection; the stop position is excluded.
display(df.iloc[0:5, 0:4])
display(df.iloc[-3:, [0, 2, 4, 10]])
````

## 7. Data cleaning: missing values
This section parallels Lab 3 Part 2. First count missing values. Then apply different treatments based on the meaning of each field.

> Always inspect before deleting or replacing data. A cleaning decision should be justified, not automatic.

````python
missing_before = df.isna().sum().sort_values(ascending=False)
display(missing_before[missing_before > 0])
print("Rows before cleaning:", len(df))
````

````python
# Work on a copy so the original imported data remains available.
df_clean = df.copy()

# Lab 3 pattern: remove records missing a required numeric field.
df_clean = df_clean.dropna(subset=["Patient_Age"])
print("Rows after removing missing Patient_Age:", len(df_clean))
````

````python
# Lab 3 pattern: replace missing categorical values with the mode.
provider_mode = df_clean["Provider_ID"].mode()[0]
df_clean["Provider_ID"] = df_clean["Provider_ID"].fillna(provider_mode)

print("Provider mode used:", provider_mode)
print("Missing Provider_ID after replacement:", df_clean["Provider_ID"].isna().sum())
````

### Interpretation checkpoint
Dropping missing ages reduces the number of rows and may change the population represented. Replacing missing provider IDs with the mode is a demonstration of the Lab 3 technique, but operationally an “Unknown” category may be more transparent. Students should distinguish a coding exercise from a defensible production-data policy.

## 8. Data transformation
This section parallels Lab 3 Part 3: create a new column, recode a category, and calculate a grouped average.

````python
# Parse the date and create a new derived column.
df_clean["Encounter_Date"] = pd.to_datetime(df_clean["Encounter_Date"])
df_clean["Encounter_Month"] = df_clean["Encounter_Date"].dt.month_name()

# New operational measure: non-wait time within total visit-related time.
df_clean["Total_Time_Min"] = (
    df_clean["Wait_Time_Min"] + df_clean["Visit_Duration_Min"]
)

display(df_clean[[
    "Encounter_ID", "Encounter_Date", "Encounter_Month",
    "Wait_Time_Min", "Visit_Duration_Min", "Total_Time_Min"
]].head())
````

````python
# Lab 3 pattern: convert a category to a binary indicator.
df_clean["Completed_Flag"] = df_clean["Visit_Status"].map({
    "Completed": 1,
    "Cancelled": 0,
    "No Show": 0
})

display(df_clean[["Visit_Status", "Completed_Flag"]].head(10))
````

````python
# Lab 3 pattern: average measure by category.
average_cost_by_service = (
    df_clean.groupby("Service_Line")["Cost_CAD"]
            .mean()
            .sort_values(ascending=False)
            .round(2)
)
display(average_cost_by_service)
````

## 9. Filtering and export
This section parallels Lab 3 Part 4: filter records above a threshold, inspect the last five records, and export the result to a new CSV.

````python
high_wait = df_clean.loc[
    df_clean["Wait_Time_Min"] > 45,
    [
        "Encounter_ID", "Facility", "Service_Line",
        "Wait_Time_Min", "Visit_Status", "Cost_CAD"
    ]
]

display(high_wait.tail(5))
print("Filtered rows:", len(high_wait))
````

````python
output_file = "filtered_healthcare_FirstLastName.csv"
high_wait.to_csv(output_file, index=False)
print("Exported:", output_file)
````
